# Data Understanding

| Column Name | Description |
| :--- | :--- |
| **transaction_id** | Unique identifier assigned to each transaction. |
| **transaction_amount** | The monetary value or amount of the transaction. |
| **login_attempts** | Number of login attempts made prior to or during the session. |
| **device_risk_score** | Quantified risk score associated with the user's device. |
| **transfer_frequency** | Frequency of money transfers made by the account. |
| **anomaly_score** | Calculated score indicating how anomalous or unusual the transaction pattern is. |
| **account_age_days** | The age of the bank account measured in days. |
| **transaction_time_hour** | Hour of the day when the transaction occurred (0 to 23 format). |
| **failed_transactions_last_30d** | Number of failed transactions recorded in the past 30 days. |
| **avg_monthly_balance** | Average monthly balance maintained in the account. |
| **daily_transaction_count** | Total number of transactions performed on that specific day. |
| **geo_distance_km** | Geographic distance in kilometers from the user's usual location or previous transaction. |
| **session_duration_minutes** | Total duration of the user's active session in minutes. |
| **transaction_velocity_score** | Score measuring the speed and frequency of transactions over a short window. |
| **payment_channel** | The channel used for the payment (e.g., POS Terminal, Web Banking, ATM, Mobile App). |
| **authentication_type** | The authentication method used for verification (e.g., OTP, Biometric). |
| **card_present_flag** | Binary flag: `1` if the physical card was present, `0` otherwise. |
| **international_transaction_flag** | Binary flag: `1` if it is an international transaction, `0` if domestic. |
| **suspicious_ip_flag** | Binary flag: `1` if the transaction originated from a flagged/suspicious IP address, `0` otherwise. |
| **fraud_flag** | Target label: `True` if the transaction is fraudulent, `False` if legitimate. |

# Data Cleaning

- In depth check for categorical columns

```python
# 1. Convert the column 'transaction_amount' to numeric values.
# Any valid number will be converted, and invalid will automatically turn into NaN instead of crashing.

converted_to_numeric = pd.to_numeric(
    df["transaction_amount"].astype(str).str.strip(), errors="coerce")

# 2. Find and select the rows where the conversion failed (turned into NaN)
error_rows = df[converted_to_numeric.isna()]

# 3. Display these rows to inspect the invalid values
print("Rows containing non-numeric values (errors):")
print(error_rows[["transaction_amount"]])
```

```python
# Convert 'transaction_amount' column from object to float 
def clean_transaction_amount(x):

    if isinstance(x, (int, float)) or pd.isna(x):
        return x

    return x.replace(' EGP', '').strip()

df['transaction_amount'] = df['transaction_amount'].apply(clean_transaction_amount).astype('float')
df['transaction_amount']
```

```python
# To drop duplicates

df = df.drop_duplicates()

# To reset index

df = df.reset_index(drop=True)
```

```python
# To drop missing values raws in 'device_risk_score' column --> very low Nan percentage

df.dropna(subset= 'device_risk_score', inplace= True, ignore_index= True)
```

# Feature Engineering

- Create Account_Age_Group column from account_age_days column

```python
# Define a function to categorize account age into groups, then apply it to create a new column
def account_groups(s):
    # If the account age is 30 days or less
    if s <= 30:
        return 'New Account'
    # If it's less than or equal to 365
    elif s <= 365:
        return 'Medium-term Account'
    # If it's greater than 365
    else:
        return 'Mature Account'
    

df['Account_Age_Group'] = df['account_age_days'].apply(account_groups)
df['Account_Age_Group']
```

```python
df.head()
```

- Create Amount_Category column from transaction_amount column

```python
# Define a function to categorize transaction amounts into tiers, then apply it to the dataset

def trans_amount(x):
   
    if x < 50:
        return 'Micro'
    elif x < 200:
        return 'Low'
    elif x < 1000:
            return 'Medium'
    elif x < 5000:
            return 'High'
    else:
        return 'Very High'
    

df['Amount_Category'] = df['transaction_amount'].apply(trans_amount)
df['Amount_Category']
```

### - How do total transaction volumes and fraud counts vary across different payment channel?

```python
# Calculate the total number of transactions and the sum of fraud cases per payment channel 
fraud_by_channel = (
    df.groupby('payment_channel')['fraud_flag'].agg(['count', 'sum']).reset_index()
)
fraud_by_channel.columns = [
    'Payment_Channel',
    'Total_Transactions',
    'Fraud_Count',
]

fraud_by_channel = fraud_by_channel.sort_values(
    by="Fraud_Count", ascending = False
).reset_index(drop=True)

fraud_by_channel
```

```python
# Create a bar chart to visualize the count of fraud cases across different payment channels
fig = px.bar(
    fraud_by_channel,
    x="Payment_Channel",
    y="Fraud_Count",
    title="Fraud Count by Payment Channel",
    text_auto=True,
    color="Payment_Channel",    
)

fig.update_layout(
    xaxis_title="The Payment Channel",  
    yaxis_title="Number of The Transactions")
fig.show()
```

### - Which Payment Channel exhibit the highest relative fraud rates?

```python
# calcualte the fraud_percentage to the Total_Transactions by_channel
fraud_by_channel["Fraud_Percentage"] = (
    (fraud_by_channel["Fraud_Count"] / fraud_by_channel["Total_Transactions"]
) * 100 ).round(2)

fraud_by_channel = fraud_by_channel.sort_values(
    by="Fraud_Percentage", ascending=False
).reset_index(drop=True)

fraud_by_channel
```

```python
# Create a grouped bar chart to visualize both metrics together along with the fraud percentage on hover
fig = px.bar(
    fraud_by_channel,
    x="Payment_Channel",
    y=["Total_Transactions", "Fraud_Count"],
    barmode="group",
    title="Total Transactions vs. Fraud Count per Payment Channel (with % Rate)",
    text_auto=True, 
    hover_data=["Fraud_Percentage"],
    labels={
        "variable": "Metric Type"
    }, 
)

fig.update_layout(
    xaxis_title="The Payment Channel",  
    yaxis_title="Number of The Transactions")

fig.show()
```

```python
# Create a pie chart to visualize the percentage and distribution of fraud cases across payment channels
fig_pie = px.pie(
    fraud_by_channel,
    names="Payment_Channel",
    values="Fraud_Count", 
    title="Fraud Distribution by Payment Channel",
    hover_data=["Total_Transactions"
                ,"Fraud_Percentage"],
)

fig_pie.update_traces(textposition="inside", textinfo="percent+label")

fig_pie.show()
```

### - Is there a relationship between transaction_time_hour and fraud_flag? Specifically, which hours of the day experience the highest rate of fraudulent transactions?

```python
# Calculate the percentage of fraud cases for each hour of the day and sort them in descending order
fraud_rate_by_hour = (
    (df.groupby("transaction_time_hour")["fraud_flag"].mean()* 100).sort_values(ascending =False)
).reset_index().round(2)


fraud_rate_by_hour
```

```python
# Sort fraud rate data chronologically by hour and create an interactive line chart with markers to show the hourly fraud trend
df_sorted = fraud_rate_by_hour.sort_values(
    by="transaction_time_hour", ascending=True
)
# Create an interactive line chart to visualize the fraud rate trend by the hour of the day with markers and text labels
fig = px.line(
    data_frame= df_sorted,
    x="transaction_time_hour",
    y="fraud_flag",
    markers=True, 
    title="Fraud Rate Trend by Hour of Day",
    text="fraud_flag",
)


fig.update_traces(
    textposition="bottom right", line=dict(color="#d9534f", width=2)
)

fig.update_layout(
    xaxis_title="Transaction Time Per Hour",  
    yaxis_title="Fraud Rate (%)",
)
fig.show()
```

### - Is there a specific transaction amount threshold where the fraud rate spikes significantly?

```python
# Group data by transaction amount category, calculate total transactions and fraud count using named aggregation, and sort descending
fraud_rate_by_Amount_Category = (
    df.groupby('Amount_Category')['fraud_flag'].agg(Total_Transactions="count", Fraud_Count="sum").reset_index()
)

fraud_rate_by_Amount_Category = fraud_rate_by_Amount_Category.sort_values(
    by="Fraud_Count", ascending = False
).reset_index(drop=True)

fraud_rate_by_Amount_Category
```

```python
# Create a bar chart to visualize fraud count distribution by transaction amount category with a red color scale
fig_bar = px.bar(
    fraud_rate_by_Amount_Category,
    x="Amount_Category",
    y="Fraud_Count",
    title="Fraud Distribution by Amount of The Transaction Category",
    text_auto=True,
    color="Fraud_Count", 
    hover_data=["Total_Transactions"],
    color_continuous_scale="Reds",
)

fig_bar.update_layout(
    xaxis_title="Amount of The Transaction Category",  
    yaxis_title="Fraud Count Transactions"
),
fig_bar.show()
```

```python
# Group data by amount category, calculate total transactions, fraud count, and precise fraud rate percentage using custom lambda aggregation
fraud_rate_by_Amount_Category = (
    df.groupby('Amount_Category')['fraud_flag'].agg(Total_Transactions="count",
                                                     Fraud_Count="sum",
                                                        Fraud_Rate= lambda x: (x.mean() * 100).round(2)).reset_index()
)

fraud_rate_by_Amount_Category = fraud_rate_by_Amount_Category.sort_values(
    by="Fraud_Rate", ascending = False
).reset_index(drop=True)

fraud_rate_by_Amount_Category
```

```python
# Create an interactive bubble chart to analyze fraud rate by amount category, using bubble size for total transactions and color for fraud rate intensity
fig_scatter = px.scatter(
    fraud_rate_by_Amount_Category,
    x="Amount_Category",
    y="Fraud_Rate",
    size="Total_Transactions", # Bubble size represents the total volume of transactions
    color="Fraud_Rate",  # Color gradient based on the fraud rate percentage
    color_continuous_scale="Reds",
    hover_data=["Fraud_Count"],
    title="Fraud Rate by Amount Category (Bubble Chart)",
    text="Fraud_Rate",
)

fig_scatter.update_traces(
    textposition="top center"
)
fig_scatter.update_layout(
    xaxis_title="Transaction Amount Category",  
    yaxis_title="Fraud Rate (%)",
)
fig_scatter.show()
```
